## 核心概念說明

AI 導入規劃的重點不是直接開始訓練模型，而是先把「可行性與價值潛力」轉換成可執行的專案計畫。

在本章節中，可將 AI 導入規劃拆成四個實務面向：

1. **情境與目的導向**：先確認產業情境、業務問題與預期效益，例如零售業可能重視銷售預測，製造業可能重視設備預測維護。
2. **資源盤點**：檢查資料品質、人力能力、硬體資源、預算與工具是否足夠。
3. **任務分解與時程設計**：使用 WBS、甘特圖或 Sprint，把大型 AI 專案拆成可追蹤的小任務。
4. **彈性調整**：AI 專案常遇到資料不足、需求變動或模型效果不佳，因此規劃必須保留緩衝與迭代空間。

接下來的程式練習會用簡化資料示範如何做專案排序、資源評估與時程視覺化。


```python
# ── 示範：AI 專案優先順序評估 ──────────────────────────
# 這段程式碼示範如何用加權評分矩陣，比較不同 AI 專案的商業價值、技術可行性、資料成熟度與組織準備度。

import pandas as pd

projects = pd.DataFrame({
    "專案": ["銷售預測", "客服文字分類", "設備預測維護"],
    "商業價值": [9, 7, 8],
    "技術可行性": [8, 9, 6],
    "資料成熟度": [7, 8, 5],
    "組織準備度": [8, 7, 6],
    "導入成本": [5, 4, 8]
})

weights = {
    "商業價值": 0.30,
    "技術可行性": 0.25,
    "資料成熟度": 0.20,
    "組織準備度": 0.15,
    "導入成本": -0.10
}

projects["總分"] = 0
for col, weight in weights.items():
    projects["總分"] += projects[col] * weight

projects = projects.sort_values("總分", ascending=False).reset_index(drop=True)

print("AI 導入專案優先順序：")
display(projects)

best_project = projects.loc[0, "專案"]
print(f"建議優先導入專案：{best_project}")
```

## 資源盤點與配置

AI 導入專案需要的資源不只包含模型與程式，也包含跨部門協作條件。常見資源可分為：

- **人力資源**：專案負責人、資料工程師、AI 模型開發者、系統架構師、測試工程師、業務顧問、UI/UX 設計師。
- **硬體資源**：GPU 伺服器、雲端運算、本地工作站、資料儲存空間。
- **開發工具**：Python、scikit-learn、版本控制、任務追蹤工具、資料處理工具。
- **資料條件**：資料來源、資料品質、欄位一致性、標註流程、隱私與法規要求。

在導入規劃中，資源不足不一定代表專案不可行，但可能代表需要先安排資料整備、教育訓練、外部採購或時程調整。


```python
# ── 示範：資源缺口分析 ───────────────────────────────
# 這段程式碼示範如何比較專案需要的資源與企業現有資源，找出可能造成時程延誤的缺口。

import pandas as pd

required = pd.DataFrame({
    "資源": ["資料工程師", "AI 模型開發者", "系統架構師", "測試工程師", "GPU 數量"],
    "需求數量": [2, 2, 1, 1, 1]
})

available = pd.DataFrame({
    "資源": ["資料工程師", "AI 模型開發者", "系統架構師", "測試工程師", "GPU 數量"],
    "現有數量": [1, 2, 0, 1, 0]
})

resource_plan = required.merge(available, on="資源")
resource_plan["缺口"] = resource_plan["需求數量"] - resource_plan["現有數量"]
resource_plan["狀態"] = resource_plan["缺口"].apply(lambda x: "需補強" if x > 0 else "足夠")

print("資源盤點結果：")
display(resource_plan)

missing = resource_plan[resource_plan["缺口"] > 0]
print("需要優先處理的資源缺口：")
display(missing[["資源", "缺口"]])
```

## WBS、甘特圖與 Sprint

AI 專案通常包含高度不確定性，因此需要把大型目標拆解成可管理的工作單元。

**WBS 工作分解結構**可以把「導入 AI 系統」拆成資料準備、模型開發、測試驗證與部署上線等任務。

**甘特圖**適合呈現任務開始時間、結束時間與重疊區段，方便跨部門溝通。

**Sprint 短衝**適合 AI 原型開發，因為模型效果、資料品質與需求常需要反覆驗證。短衝規劃通常會根據任務價值、風險與工作量安排優先順序。


```python
# ── 實際應用：建立簡化甘特圖 ────────────────────────────
# 這段程式碼使用 pandas 與 matplotlib 建立 AI 導入專案的簡化甘特圖，呈現各任務的時程安排。

import pandas as pd
import matplotlib.pyplot as plt

schedule = pd.DataFrame({
    "任務": ["需求訪談", "資料盤點", "資料清理", "模型原型", "驗證測試", "部署規劃"],
    "開始週": [1, 1, 3, 5, 8, 10],
    "工期週數": [2, 2, 3, 4, 3, 2],
    "負責角色": ["專案負責人", "資料工程師", "資料工程師", "AI 模型開發者", "測試工程師", "系統架構師"]
})

schedule["結束週"] = schedule["開始週"] + schedule["工期週數"]

fig, ax = plt.subplots(figsize=(10, 5))
for i, row in schedule.iterrows():
    ax.barh(row["任務"], row["工期週數"], left=row["開始週"], color="#4C78A8")
    ax.text(row["結束週"] + 0.1, i, row["負責角色"], va="center")

ax.set_xlabel("週次")
ax.set_ylabel("任務")
ax.set_title("AI 導入專案簡化甘特圖")
ax.grid(axis="x", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.show()

display(schedule)
```
