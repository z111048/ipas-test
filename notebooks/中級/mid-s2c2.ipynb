## 核心概念說明

機率分佈用來描述隨機變數可能出現的值，以及每個值或每段區間出現的機率。

### 1. 離散型機率分佈

離散型隨機變數的取值是有限或可數的，例如：

- 一次品質檢測是否合格：0 或 1
- 10 次廣告曝光中有幾次被點擊：0 到 10
- 每分鐘客服中心接到幾通電話：0、1、2、3 ...

常見模型包含伯努利分佈、二項分佈與泊松分佈。

### 2. 連續型機率分佈

連續型隨機變數可以取某個範圍內的任意實數，例如：

- 身高
- 體重
- 等待時間
- 零件壽命

常見模型包含常態分佈、指數分佈與卡方分佈。

### 3. 期望值與變異數

- 期望值代表長期平均結果。
- 變異數代表資料分散程度。
- 標準差是變異數的平方根，單位與原始資料相同，較容易解讀。


```python
# ── 示範：離散型分佈 ────────────────────────────────
# 這段程式碼示範伯努利分佈、二項分佈與泊松分佈，分別對應單次成功失敗、多次成功次數，以及固定時間內事件發生次數。

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import bernoulli, binom, poisson

np.random.seed(42)

# 伯努利分佈：單次廣告是否被點擊
click_probability = 0.25
bernoulli_samples = bernoulli.rvs(click_probability, size=1000)

# 二項分佈：每 20 次廣告曝光中，被點擊的次數
n_ads = 20
binomial_samples = binom.rvs(n=n_ads, p=click_probability, size=1000)

# 泊松分佈：每小時客服來電數，平均每小時 6 通
lambda_calls = 6
poisson_samples = poisson.rvs(mu=lambda_calls, size=1000)

summary = pd.DataFrame({
    '分佈': ['伯努利', '二項', '泊松'],
    '樣本平均': [bernoulli_samples.mean(), binomial_samples.mean(), poisson_samples.mean()],
    '樣本變異數': [bernoulli_samples.var(), binomial_samples.var(), poisson_samples.var()],
    '理論期望值': [click_probability, n_ads * click_probability, lambda_calls],
    '理論變異數': [click_probability * (1 - click_probability), n_ads * click_probability * (1 - click_probability), lambda_calls]
})

print(summary.round(3))

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
axes[0].hist(bernoulli_samples, bins=[-0.5, 0.5, 1.5], edgecolor='black')
axes[0].set_title('Bernoulli: 是否點擊')
axes[0].set_xticks([0, 1])

axes[1].hist(binomial_samples, bins=range(0, n_ads + 2), edgecolor='black')
axes[1].set_title('Binomial: 20 次曝光點擊數')

axes[2].hist(poisson_samples, bins=range(0, max(poisson_samples) + 2), edgecolor='black')
axes[2].set_title('Poisson: 每小時來電數')

for ax in axes:
    ax.set_xlabel('取值')
    ax.set_ylabel('次數')

plt.tight_layout()
plt.show()
```

## 連續型分佈的觀察重點

連續型分佈不關心某個單一數值的機率，而是關心某個區間內的機率。

例如等待時間是連續型變數，顧客剛好等待 `3.000000` 分鐘的機率幾乎沒有實務意義；比較合理的問法是：

- 等待時間小於 3 分鐘的機率是多少？
- 等待時間介於 3 到 5 分鐘的機率是多少？

常態分佈常用於描述集中且對稱的資料，例如測量誤差、身高或考試分數。指數分佈常用於描述等待時間，例如來電間隔、設備故障間隔或服務到達間隔。


```python
# ── 示範：連續型分佈 ────────────────────────────────
# 這段程式碼示範常態分佈與指數分佈，並計算特定區間的機率。

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm, expon

np.random.seed(42)

# 常態分佈：考試分數，平均 70 分，標準差 10 分
mu = 70
sigma = 10
scores = norm.rvs(loc=mu, scale=sigma, size=2000)

# 指數分佈：客服來電間隔，平均等待 5 分鐘，因此 scale = 5
mean_wait = 5
wait_times = expon.rvs(scale=mean_wait, size=2000)

# 區間機率
prob_score_60_80 = norm.cdf(80, loc=mu, scale=sigma) - norm.cdf(60, loc=mu, scale=sigma)
prob_wait_under_3 = expon.cdf(3, scale=mean_wait)

print(f'考試分數介於 60 到 80 分的機率：約 {prob_score_60_80:.3f}')
print(f'等待時間小於 3 分鐘的機率：約 {prob_wait_under_3:.3f}')

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(scores, bins=30, density=True, alpha=0.7, edgecolor='black')
x_score = np.linspace(30, 110, 300)
axes[0].plot(x_score, norm.pdf(x_score, loc=mu, scale=sigma), linewidth=2)
axes[0].set_title('Normal: 考試分數')
axes[0].set_xlabel('分數')
axes[0].set_ylabel('密度')

axes[1].hist(wait_times, bins=30, density=True, alpha=0.7, edgecolor='black')
x_wait = np.linspace(0, 30, 300)
axes[1].plot(x_wait, expon.pdf(x_wait, scale=mean_wait), linewidth=2)
axes[1].set_title('Exponential: 等待時間')
axes[1].set_xlabel('分鐘')
axes[1].set_ylabel('密度')

plt.tight_layout()
plt.show()
```

## 用資料判斷可能的分佈模型

知道各種分佈的特性後，實務上的問題是反過來的：拿到一份資料，該用哪個分佈描述它？以下建立一份模擬營運資料（廣告點擊、每日訂單數、客服等待時間），依資料型態挑選合適的分佈模型並做摘要。


```python
# ── 實際應用：用資料判斷可能的分佈模型 ───────────────────────
# 這段程式碼建立一份模擬營運資料，包含廣告點擊、每日訂單數與客服等待時間，並依資料型態選擇合適的分佈模型進行摘要。

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)

n_days = 120
ad_exposures = np.random.randint(80, 160, size=n_days)
click_rate = 0.08
clicks = np.random.binomial(n=ad_exposures, p=click_rate)
orders = np.random.poisson(lam=12, size=n_days)
wait_minutes = np.random.exponential(scale=4, size=n_days)

business_data = pd.DataFrame({
    'ad_exposures': ad_exposures,
    'clicks': clicks,
    'orders': orders,
    'wait_minutes': wait_minutes
})

business_data['observed_ctr'] = business_data['clicks'] / business_data['ad_exposures']

print('資料前五列：')
print(business_data.head())

print('\n營運指標摘要：')
summary = business_data.agg({
    'ad_exposures': ['mean', 'std'],
    'clicks': ['mean', 'std'],
    'orders': ['mean', 'std'],
    'wait_minutes': ['mean', 'std'],
    'observed_ctr': ['mean', 'std']
})
print(summary.round(3))

model_mapping = pd.DataFrame({
    '欄位': ['clicks', 'orders', 'wait_minutes'],
    '資料意義': ['固定曝光次數中的點擊數', '每天訂單數', '客服等待時間'],
    '可能分佈': ['二項分佈', '泊松分佈', '指數分佈'],
    '判斷理由': ['成功次數，且每次曝光可視為點擊或未點擊', '固定期間內事件發生次數', '事件發生前的等待時間']
})
print('\n模型選擇建議：')
print(model_mapping)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
business_data['clicks'].hist(ax=axes[0], bins=15, edgecolor='black')
axes[0].set_title('每日點擊數')
axes[0].set_xlabel('點擊數')
axes[0].set_ylabel('天數')

business_data['orders'].hist(ax=axes[1], bins=15, edgecolor='black')
axes[1].set_title('每日訂單數')
axes[1].set_xlabel('訂單數')
axes[1].set_ylabel('天數')

business_data['wait_minutes'].hist(ax=axes[2], bins=15, edgecolor='black')
axes[2].set_title('客服等待時間')
axes[2].set_xlabel('分鐘')
axes[2].set_ylabel('天數')

plt.tight_layout()
plt.show()
```
