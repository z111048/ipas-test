## 核心概念說明

AI 導入評估的重點，不是先問「能不能使用 AI」，而是先判斷「是否值得導入 AI」。一個成熟的評估流程通常包含四個面向：

- **業務需求**：是否存在明確痛點？是否能改善 KPI，例如效率、準確率、回應速度或成本？
- **技術可行性**：資料品質、系統整合、IT 基礎設施與 AI 技術路線是否足以支撐導入？
- **財務效益**：投入成本是否合理？ROI、NPV 與回收期是否符合組織期待？
- **組織成熟度**：資料治理、跨部門協作、人才能力、預算與責任單位是否到位？

本 Notebook 會用簡化的量化模型示範如何把這些概念轉成可比較的分數。


```python
# ── 示範：AI 專案導入優先順序評分 ────────────────────────
# 這段程式碼示範如何將業務需求、技術條件、效益與風險轉成綜合導入分數。

import numpy as np
import pandas as pd

projects = pd.DataFrame({
    "專案": ["客服自動回覆", "瑕疵影像檢測", "銷售需求預測", "內部知識問答"],
    "業務痛點分數": [9, 8, 7, 6],
    "資料品質分數": [7, 8, 6, 5],
    "IT整合分數": [8, 6, 7, 6],
    "預期效益分數": [8, 9, 7, 6],
    "風險分數": [4, 6, 5, 7]
})

weights = {
    "業務痛點分數": 0.30,
    "資料品質分數": 0.20,
    "IT整合分數": 0.15,
    "預期效益分數": 0.25,
    "風險分數": -0.10
}

projects["導入優先分數"] = sum(projects[col] * weight for col, weight in weights.items())
projects = projects.sort_values("導入優先分數", ascending=False)

print(projects[["專案", "導入優先分數"]])
```

## 成本效益分析

AI 導入通常需要模型開發、資料清洗、系統整合、授權費、教育訓練與維運成本。因此只看技術可行性不足以決策，還需要檢查財務合理性。

常見指標包括：

- **ROI 投資報酬率**：衡量每投入一單位成本可帶來多少回報。
- **NPV 淨現值**：將未來現金流折現後，判斷專案是否創造正向價值。
- **Payback Period 回收期**：估計需要多久才能回收初期投資。

若 ROI 高、NPV 為正、回收期短，通常代表專案較適合優先試點。


```python
# ── 示範：ROI、NPV 與回收期計算 ───────────────────────
# 這段程式碼示範如何用簡化財務模型評估 AI 導入是否具備投資價值。

import numpy as np

initial_investment = 1_200_000
annual_benefits = np.array([450_000, 600_000, 750_000, 850_000])
annual_operating_costs = np.array([150_000, 180_000, 200_000, 220_000])
discount_rate = 0.08

net_cash_flows = annual_benefits - annual_operating_costs
total_return = net_cash_flows.sum()
roi = (total_return - initial_investment) / initial_investment * 100
npv = sum(cash / ((1 + discount_rate) ** year) for year, cash in enumerate(net_cash_flows, start=1)) - initial_investment
payback_period = initial_investment / net_cash_flows.mean()

print(f"每年淨現金流：{net_cash_flows}")
print(f"ROI：{roi:.2f}%")
print(f"NPV：{npv:,.0f} 元")
print(f"回收期：約 {payback_period:.2f} 年")
```

```python
# ── 實際應用：組織成熟度雷達圖 ───────────────────────────
# 這段程式碼將數位成熟度、資料治理、團隊能力、跨部門協作與預算支持視覺化，協助判斷導入條件是否均衡。

import numpy as np
import matplotlib.pyplot as plt

labels = ["數位成熟度", "資料治理", "團隊能力", "跨部門協作", "預算支持"]
scores = np.array([7, 6, 5, 8, 6])

angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False)
scores_closed = np.concatenate([scores, [scores[0]]])
angles_closed = np.concatenate([angles, [angles[0]]])

plt.figure(figsize=(6, 6))
ax = plt.subplot(111, polar=True)
ax.plot(angles_closed, scores_closed, marker="o")
ax.fill(angles_closed, scores_closed, alpha=0.2)
ax.set_xticks(angles)
ax.set_xticklabels(labels)
ax.set_ylim(0, 10)
ax.set_title("AI 導入組織成熟度盤點")
plt.show()

weak_items = [label for label, score in zip(labels, scores) if score < 6]
print("需優先補強項目：", weak_items)
```
