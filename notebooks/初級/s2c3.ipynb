```python
# ── 環境設定 ────────────────────────────────────
# 載入本章節會用到的 Python 套件，並建立一組企業導入生成式 AI 的候選應用資料。

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

use_cases = pd.DataFrame({
    "方案": ["客服回覆助理", "內部知識檢索", "行銷文案生成", "供應鏈異常摘要", "合約條款檢查"],
    "業務痛點": [
        "客服等待時間過長 回覆品質不一致",
        "員工查找制度 文件 流程耗時",
        "行銷內容產出速度慢 缺乏多版本測試",
        "供應鏈資料分散 異常原因難以快速整理",
        "合約審閱耗時 法規與條款風險需要初步標記"
    ],
    "業務價值": [5, 4, 3, 4, 5],
    "技術可行性": [4, 5, 5, 3, 3],
    "資料準備度": [4, 3, 4, 3, 2],
    "風險可控度": [3, 4, 4, 3, 2]
})

print("候選 AI 導入方案：")
display(use_cases)
```

## 核心概念說明

生成式 AI 導入不是單純選一個工具，而是要從企業需求、資料、技術、人員與風險多面向評估。

常見評估構面如下：

- **需求與現狀評估**：企業是否有明確痛點？AI 是否能處理這個問題？
- **資源與基礎設施評估**：資料品質、系統架構、技術人才是否足夠？
- **階段性導入策略**：是否能先從小規模 POC 開始，再逐步擴展？
- **員工技能與文化導入**：使用者是否能理解並接受 AI 工具？
- **風險評估與管理**：是否考慮資料安全、隱私、模型偏誤、法規遵循？

在實務上，可以先建立一個評分模型，將每個候選應用依不同構面量化，協助決策者排序。


```python
# ── 示範：AI 導入評估矩陣 ────────────────────────────
# 這段程式碼示範如何把多個評估構面轉換成加權總分，找出最適合優先試點的生成式 AI 應用。

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

use_cases = pd.DataFrame({
    "方案": ["客服回覆助理", "內部知識檢索", "行銷文案生成", "供應鏈異常摘要", "合約條款檢查"],
    "業務價值": [5, 4, 3, 4, 5],
    "技術可行性": [4, 5, 5, 3, 3],
    "資料準備度": [4, 3, 4, 3, 2],
    "風險可控度": [3, 4, 4, 3, 2]
})

weights = {
    "業務價值": 0.35,
    "技術可行性": 0.25,
    "資料準備度": 0.25,
    "風險可控度": 0.15
}

use_cases["導入優先分數"] = sum(use_cases[col] * weight for col, weight in weights.items())
ranked = use_cases.sort_values("導入優先分數", ascending=False)

print("AI 導入優先順序：")
display(ranked)

plt.figure(figsize=(8, 4))
plt.bar(ranked["方案"], ranked["導入優先分數"])
plt.title("生成式 AI 導入優先分數")
plt.ylabel("加權分數")
plt.xticks(rotation=30, ha="right")
plt.ylim(0, 5)
plt.tight_layout()
plt.show()
```

## 從評估到試點

完成初步排序後，企業通常不會立刻全面導入，而是先進行 POC，也就是概念驗證。

POC 階段常見檢查項目：

- 模型是否能在真實業務資料上產生有用結果。
- 是否能降低錯誤率、節省時間或提升服務品質。
- 使用者是否能順利把 AI 工具放入既有流程。
- 資料安全、隱私與法規要求是否可被滿足。

接著可用簡化 ROI 評估：

**ROI =（預期效益 - 導入成本）/ 導入成本**

ROI 不是唯一標準，但能幫助企業比較不同導入方案的短期效益與資源投入。


```python
# ── 示範：用 TF-IDF 模擬痛點與應用場景匹配 ─────────────────
# 大型語言模型常用語意向量理解文本相似度；本練習用 Colab 預裝的 TF-IDF 建立輕量版相似度示範，協助判斷企業痛點適合哪一類 AI 應用。

import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

pain_point = "客服等待時間太長，回答內容不一致，主管希望降低客服人力負擔"

ai_scenarios = pd.DataFrame({
    "應用場景": ["客服回覆助理", "內部知識檢索", "行銷文案生成", "合約條款檢查"],
    "場景描述": [
        "自動整理客戶問題 產生一致且快速的客服回覆",
        "協助員工搜尋內部文件 制度 流程與知識庫",
        "快速產生廣告文案 社群貼文與多版本行銷內容",
        "初步標記合約條款 法規風險與審閱重點"
    ]
})

texts = [pain_point] + ai_scenarios["場景描述"].tolist()
vectorizer = TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 4))
tfidf = vectorizer.fit_transform(texts)

similarities = cosine_similarity(tfidf[0:1], tfidf[1:]).flatten()
ai_scenarios["痛點匹配分數"] = similarities
result = ai_scenarios.sort_values("痛點匹配分數", ascending=False)

print("企業痛點：", pain_point)
print("最相近的 AI 應用場景：")
display(result)
```

## 實際應用：建立導入路線圖

企業可將候選 AI 應用分成三類：

- **立即試點**：高價值、低難度、資料較完整、風險可控。
- **準備後導入**：價值高，但資料、人才或系統尚未成熟。
- **暫緩推動**：技術難度高、風險高，或短期效益不明確。

這種分類能呼應教材中的分階段策略：先試點驗證，再擴大整合，最後進入成熟階段的全面創新。


```python
# ── 實際應用：AI 導入路線圖分類 ─────────────────────────
# 這段程式碼根據導入優先分數與風險可控度，自動把 AI 應用分成不同導入階段，模擬企業進行生成式 AI 導入規劃的決策流程。

import pandas as pd

projects = pd.DataFrame({
    "方案": ["客服回覆助理", "內部知識檢索", "行銷文案生成", "供應鏈異常摘要", "合約條款檢查"],
    "導入優先分數": [4.05, 4.05, 4.00, 3.30, 3.20],
    "風險可控度": [3, 4, 4, 3, 2],
    "資料準備度": [4, 3, 4, 3, 2]
})

def classify_stage(row):
    if row["導入優先分數"] >= 4.0 and row["風險可控度"] >= 3:
        return "立即試點"
    if row["導入優先分數"] >= 3.3 and row["資料準備度"] >= 3:
        return "準備後導入"
    return "暫緩推動"

projects["建議導入階段"] = projects.apply(classify_stage, axis=1)

print("生成式 AI 導入路線圖：")
display(projects.sort_values(["建議導入階段", "導入優先分數"], ascending=[True, False]))

summary = projects["建議導入階段"].value_counts()
print("\n各階段方案數量：")
print(summary)
```

```python
# ── 🧪 自我測驗 ──────────────────────────────────
# 請完成下方 TODO 填空，實作 ROI 計算與導入建議判斷。這個練習對應教材中的成本效益評估、POC 驗證與階段性導入規劃。

import pandas as pd

poc_results = pd.DataFrame({
    "方案": ["客服回覆助理", "內部知識檢索", "合約條款檢查"],
    "導入成本": [120000, 90000, 180000],
    "預期年度效益": [210000, 135000, 200000],
    "POC成功率": [0.82, 0.76, 0.58]
})

# TODO 1: 計算 ROI，公式為（預期年度效益 - 導入成本）/ 導入成本
poc_results["ROI"] = (poc_results["預期年度效益"] - poc_results["導入成本"]) / poc_results["導入成本"]

# TODO 2: 若 ROI >= 0.5 且 POC成功率 >= 0.75，建議為「擴大導入」，否則為「重新評估」
def make_decision(row):
    if row["ROI"] >= 0.5 and row["POC成功率"] >= 0.75:
        return "擴大導入"
    return "重新評估"

poc_results["導入建議"] = poc_results.apply(make_decision, axis=1)

print(poc_results[["方案", "ROI", "POC成功率", "導入建議"]])

# Expected:
# 客服回覆助理 ROI 約 0.75，導入建議為 擴大導入
# 內部知識檢索 ROI 約 0.50，導入建議為 擴大導入
# 合約條款檢查 ROI 約 0.11，導入建議為 重新評估
```
